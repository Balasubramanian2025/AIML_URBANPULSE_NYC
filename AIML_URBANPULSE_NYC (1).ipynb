```python
# Create a working copy
df_clean = df.copy()

# Fill missing categorical values
df_clean['borough'] = (
    df_clean['borough'].fillna('UNKNOWN')
)

df_clean['descriptor'] = (
    df_clean['descriptor'].fillna('Not Provided')
)

df_clean['status'] = (
    df_clean['status'].fillna('Unknown')
)

# Check remaining missing values
display(df_clean.isnull().sum())
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate Rows:", duplicate_count)

df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```

```python
print(
    "Duplicate Complaint IDs:",
    df_clean['unique_key'].duplicated().sum()
)
```

```python
df_clean['created_date'] = pd.to_datetime(
    df_clean['created_date'],
    errors='coerce'
)

df_clean['closed_date'] = pd.to_datetime(
    df_clean['closed_date'],
    errors='coerce'
)

print(df_clean[['created_date', 'closed_date']].dtypes)
```

```python
df_clean['resolution_hours'] = (
    df_clean['closed_date'] -
    df_clean['created_date']
).dt.total_seconds() / 3600

# Remove invalid negative durations
df_clean.loc[
    df_clean['resolution_hours'] < 0,
    'resolution_hours'
] = np.nan

display(df_clean['resolution_hours'].describe())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load NYC 311 dataset
url = "https://data.cityofnewyork.us/resource/erm2-nwe9.csv?$limit=10000"

df = pd.read_csv(url)

# Create cleaned DataFrame
df_clean = df.copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Convert date columns
df_clean['created_date'] = pd.to_datetime(
    df_clean['created_date'],
    errors='coerce'
)

df_clean['closed_date'] = pd.to_datetime(
    df_clean['closed_date'],
    errors='coerce'
)

# Create date and time columns
df_clean['date'] = df_clean['created_date'].dt.date
df_clean['day_name'] = df_clean['created_date'].dt.day_name()
df_clean['hour'] = df_clean['created_date'].dt.hour
df_clean['year_month'] = df_clean['created_date'].dt.to_period('M')

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df_clean.shape)

display(df_clean.head())
```

```python
print(df_clean.shape)
print(df_clean.columns.tolist())

display(
    df_clean[
        ['created_date', 'date', 'day_name', 'hour', 'year_month']
    ].head()
)
```

```python
# Daily complaints
daily_counts = df_clean.groupby('date').size()

# Weekday complaints
weekday_counts = df_clean['day_name'].value_counts()

# Hourly complaints
hourly_counts = df_clean['hour'].value_counts()

# Monthly complaints
monthly_counts = df_clean.groupby('year_month').size()

print("Daily Complaints:")
print(daily_counts.head())

print("\nWeekday Complaints:")
print(weekday_counts)

print("\nHourly Complaints:")
print(hourly_counts.sort_index())

print("\nMonthly Complaints:")
print(monthly_counts)
```

```python
heatmap_data = pd.crosstab(
    df_clean['day_name'],
    df_clean['hour']
)

heatmap_data = heatmap_data.reindex(
    index=day_order,
    columns=range(24),
    fill_value=0
)

plt.figure(figsize=(15, 6))

sns.heatmap(
    heatmap_data,
    cmap='YlOrRd',
    annot=True,
    fmt='d'
)

plt.title("NYC 311 Complaints: Weekday vs Hour")
plt.xlabel("Hour")
plt.ylabel("Weekday")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from google.colab import files

columns = [
    'unique_key', 'created_date', 'closed_date',
    'agency', 'agency_name', 'complaint_type',
    'descriptor', 'borough', 'status',
    'latitude', 'longitude', 'resolution_hours'
]

# Recalculate if necessary
if 'resolution_hours' not in df_clean.columns:
    df_clean['resolution_hours'] = (
        pd.to_datetime(df_clean['closed_date'], errors='coerce') -
        pd.to_datetime(df_clean['created_date'], errors='coerce')
    ).dt.total_seconds() / 3600

df_mysql = df_clean[columns].copy()
df_mysql = df_mysql.dropna(subset=['unique_key'])
df_mysql = df_mysql.drop_duplicates(subset=['unique_key'])

# Format date columns for MySQL
for col in ['created_date', 'closed_date']:
    df_mysql[col] = pd.to_datetime(
        df_mysql[col], errors='coerce'
    ).dt.strftime('%Y-%m-%d %H:%M:%S')

# Represent missing values as MySQL NULL
df_mysql = df_mysql.astype(object).where(
    pd.notna(df_mysql), r'\N'
)

df_mysql.to_csv(
    'nyc_311_mysql.csv',
    index=False
)

files.download('nyc_311_mysql.csv')
```
